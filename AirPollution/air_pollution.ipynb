```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("global_air_pollution_dataset.csv")
```

```python
top5 = df.sort_values(by="AQI Value", ascending=False).head(5)
top5
```

```python
top5["City_Country"] = top5["City"] + " (" + top5["Country"] + ")"
```

```python
plt.figure()
plt.bar(top5["City_Country"], top5["AQI Value"])
plt.xlabel("Shahar (Davlat)")
plt.ylabel("AQI Value")
plt.title("Eng yomon 5 ta shahar (AQI bo‘yicha)")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
best5 = df.sort_values(by="AQI Value", ascending=True).head(5)
best5
```

```python
best5["City_Country"] = best5["City"] + " (" + best5["Country"] + ")"
```

```python
plt.figure()
plt.bar(best5["City_Country"], best5["AQI Value"])
plt.xlabel("Shahar (Davlat)")
plt.ylabel("AQI Value")
plt.title("Eng yaxshi 5 ta shahar (AQI bo‘yicha)")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
worst_city = df.sort_values(by="AQI Value", ascending=False).head(1)
worst_city
```

```python
city_name = worst_city["City"].values[0]
country_name = worst_city["Country"].values[0]

title_name = f"{city_name} ({country_name})"
```

```python
labels = [
    "CO AQI Value",
    "Ozone AQI Value",
    "NO2 AQI Value",
    "PM2.5 AQI Value"
]

values = [
    worst_city["CO AQI Value"].values[0],
    worst_city["Ozone AQI Value"].values[0],
    worst_city["NO2 AQI Value"].values[0],
    worst_city["PM2.5 AQI Value"].values[0]
]
```

```python
plt.figure()
plt.pie(values, labels=labels, autopct='%1.1f%%', startangle=90)
plt.title(f"Eng yomon shahar bo‘yicha AQI tarkibi\n{title_name}")
plt.tight_layout()
plt.show()
```
